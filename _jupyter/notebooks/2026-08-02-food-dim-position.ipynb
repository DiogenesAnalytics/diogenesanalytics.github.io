## Introduction
Food is essential for life. Every human civilization has depended upon a reliable supply of energy and nutrients, and throughout history the acquisition, production, storage, and distribution of food have shaped settlement patterns, economic systems, technological development, and political power. Despite its obvious importance, discussions of food typically focus on its biological, agricultural, economic, or nutritional roles.

This article asks a different question:

> **What role does food play in an agent's position within the game of life?**

Within the Position framework, an agent's position is the collection of information that determines the threats it faces, the opportunities it can exploit, and the future states it can potentially reach. Position is therefore not merely a description of an agent's current circumstances; it is the informational state from which all future decisions and outcomes emerge.

The central hypothesis of this article is that **food constitutes a fundamental dimension of position**. There exist classes of threats and opportunities whose outcomes cannot be determined without information about an agent's relationship to food. Consequently, any sufficiently complete description of an agent's position must include a food dimension.

Food position is not simply a measure of the food currently available to an agent. Rather, it describes the broader state governing an agent's ability to obtain adequate nutrition across time. This includes the ability to acquire food, produce food, preserve food, store reserves, maintain nutritional quality, and withstand disruptions to supply. Two agents may possess identical present resources yet occupy different food positions if their future access to nutrition depends upon different levels of security, resilience, and productive capacity.

The purpose of this article is not to study food itself. Rather, it is to investigate why food is a fundamental dimension of position, identify the kinds of information that comprise an agent's food position, and explore how changes in that position expand or constrain the set of possible future states.

More broadly, this article serves as another case study in the methodology of identifying and analyzing the dimensions of position. If food can be understood as a dimension of position, then the same methodology can be applied to other fundamental dimensions of the human game, revealing the structure of the position space within which agents make decisions and pursue future outcomes.


## The Nature of Food Position
Food is a requirement of continued biological existence. An organism must continually acquire matter and energy from its surroundings in order to maintain itself, grow, reproduce, and perform the activities necessary to survive. For humans, food is the primary means by which this requirement is satisfied. Unlike many of the resources that improve an individual's circumstances, food cannot simply be treated as something that is desirable to possess. Its absence eventually places a direct constraint on the organism's ability to continue existing.

This makes the relationship between an individual and food an important part of that individual's position.

However, food position is not simply a measure of how much food an individual currently possesses. Consider two individuals who each have enough food for three days. The first lives in a community with reliable access to markets and has sufficient resources to purchase additional food. The second is isolated, has no means of acquiring additional supplies, and possesses no ability to produce food. Although their immediate food inventories are identical, their positions are not. The first individual has a mechanism for maintaining access to food after the existing supply is exhausted; the second does not.

The distinction is important because food is consumed. A quantity of food possessed at one moment does not remain available indefinitely. A food position therefore cannot be understood solely as an inventory. It must also incorporate the conditions under which that inventory can be replenished.

This gives food position an inherently temporal character. At any moment, an individual has some quantity of food available to them, but they also exist within a process of consumption and replenishment. Food is acquired, stored, consumed, produced, exchanged, and lost. The position of an individual therefore depends not only on what food is available now, but on what food will be available in the future and on the individual's ability to obtain it.

The same principle applies to the adequacy of food. Possessing a large quantity of something that cannot satisfy the organism's nutritional requirements does not necessarily constitute a strong food position. An adequate food position must provide the substances required to sustain biological function. Thus, quantity alone cannot completely describe the relationship between an individual and food.

We can therefore begin to distinguish several concepts that are often treated as interchangeable. **Food possession** concerns what food an individual currently controls. **Food availability** concerns what food is accessible within the individual's surrounding environment. **Food access** concerns the individual's ability to obtain that food. **Food security** concerns the reliability of that access through time. **Food position** is the broader state of the individual with respect to their present and future ability to obtain sufficient food to maintain their biological function.

This distinction also explains why food position can change without any immediate change in the amount of food an individual possesses. A disruption to transportation, a loss of income, a failed harvest, the destruction of storage, or the loss of a food supplier can all alter an individual's future access to food while leaving the current inventory unchanged. Conversely, acquiring new sources of income, developing the ability to produce food, establishing reserves, or gaining access to additional suppliers can improve food position without immediately increasing the amount of food being consumed.

Food position is therefore best understood as a **dynamic relationship between an organism and the resources and processes through which it obtains nourishment**. The food itself is only one part of that relationship. The ability to acquire, preserve, produce, exchange, and replenish food determines how the present state translates into future states.

This gives us a useful starting point for examining the food dimension of position. Rather than asking simply, *How much food does an individual have?*, we can ask a more general question:

> What is the individual's ability to maintain adequate access to food through time?

The answer to that question will require us to examine the different components of food position and the mechanisms through which those components change. Those components need not be assumed in advance. They can instead be developed by examining the ways in which an individual's relationship with food can improve, deteriorate, persist, or become vulnerable to disruption.
