## Introduction

In [The Wealth Problem][1], we formulated wealth management as a constrained optimization problem in which an agent seeks to maximize positional utility over a planning horizon. The formulation established the objective, but solving the problem requires a model of how an agent's actions change the wealth it controls and the position it occupies.

Consider an agent deciding whether to retain cash, purchase an asset, repay a liability, or spend resources on an improvement. Each action changes the agent's wealth in a different way. The consequences may extend beyond the monetary value of the resulting collection, affecting liquidity, obligations, and the actions available in the future.

To compare these decisions mathematically, we must represent the agent's wealth and describe how it changes in response to actions and environmental conditions.

Let $W_t$ denote the agent's wealth collection at time $t$, $a_t$ the action it selects, and $E_t$ the relevant environmental conditions. We seek to establish a transition function

$$
W_{t+1}=F_W(W_t,a_t,E_t)
$$

that determines the resulting wealth collection.

The objective is not merely to calculate changes in monetary wealth. We must preserve the distinctions between resources that affect how they can be used, transformed, transferred, and retained. Those distinctions matter because the resulting wealth collection influences the agent's available actions and, ultimately, its position.

This article develops a mathematical representation of wealth and establishes how actions and environmental conditions determine its transitions. The resulting model will provide a foundation for evaluating alternative wealth-management decisions in terms of their consequences for the agent's position.
