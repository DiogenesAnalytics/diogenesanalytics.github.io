## Introduction
In *Environment as a Fundamental Dimension of Position*, we argued that an agent's position depends not only on internal characteristics such as knowledge, resources, and health, but also on the environment in which those characteristics are expressed. Different environments provide different opportunities, impose different constraints, and ultimately determine different sets of reachable future positions.

Recognizing environment as a fundamental dimension of position immediately gives rise to a new strategic problem.

> **How should a rational agent acquire access to environments?**

For most individuals, environmental access is not unrestricted. The ability to live, work, invest, establish businesses, own property, or participate in another society is governed by legal systems that regulate who may enter and under what conditions. These legal mechanisms—including visas, residency programs, and citizenship pathways—determine which environments are available to the agent.

Viewed in isolation, a visa appears to be little more than a bureaucratic document. From the perspective of the Position framework, however, a visa is something far more significant. It is a mechanism that changes the set of environments in which an agent can legally operate, thereby changing the set of future positions that become attainable.

This naturally suggests another question:

> **What is the optimal visa strategy?**

At first glance, the answer appears straightforward: identify the "best" visa and pursue it. This is a pure strategy. It assumes that one pathway dominates all others and that the optimal decision is simply to commit to that pathway.

Real life is rarely so simple.

The future is uncertain. Political conditions change. Immigration laws evolve. Economic opportunities emerge and disappear. Personal goals shift over time. Under these conditions, committing to a single irreversible pathway may not maximize long-term position.

Instead, rational agents often benefit from maintaining multiple environmental pathways simultaneously, delaying irreversible commitments when appropriate, and preserving the flexibility to adapt as new information becomes available.

The problem, therefore, is not simply choosing a visa.

It is designing a strategy for acquiring and managing environmental access over time.

This article develops a theory of visa strategy from this broader perspective. Rather than treating visas as isolated documents, we treat them as strategic mechanisms for expanding an agent's future reachable state space. The objective is not to identify a universally "best" visa, but to develop principles for constructing robust environmental access strategies under uncertainty.
