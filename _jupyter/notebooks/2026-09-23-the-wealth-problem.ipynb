## Introduction
Having wealth does not end the problem of wealth.

The previous article [*Defining Wealth*][1] examined what distinguishes wealth from the many other things that can matter to an agent. Something can have value without being wealth. Something can affect an agent's position without being wealth. A capability, a piece of knowledge, or an opportunity may be valuable and may change what an agent can do without itself constituting an economically realizable resource controlled by the agent.

The investigation led to the following definition:

$$
\boxed{
\text{Wealth is an economically realizable resource that an agent can control and retain through time.}
}
$$

This definition provides the starting point for the present problem.

Wealth is not merely something that has value. It is a resource that an agent can control and retain. A house, a quantity of money, a piece of land, equipment, inventory, or an economic claim can constitute wealth because the agent can exercise control over the resource and retain it as a resource through time.

The temporal aspect is particularly important.

If an agent can retain a resource through time, then the resource can exist at one point in time and still be available at a later point. What the agent possesses at the later time therefore depends on what happened to the resources it possessed earlier.

A resource may remain under the agent's control. It may be consumed. It may be exchanged for another resource. It may be transferred to another agent. It may be used in production. It may be transformed into another form of wealth. It may be impaired or destroyed. The agent may also take actions intended to prevent some of these outcomes.

The problem therefore begins with the transition of wealth through time.

At time $t$, an agent possesses some collection of wealth resources. At a later time $t+1$, it possesses another collection:

$$
W_t\rightarrow W_{t+1}.
$$

At this stage, $W_t$ need not be understood as a single numerical quantity. It represents the wealth resources controlled by the agent at time $t$. Those resources may be heterogeneous, and the transition between $W_t$ and $W_{t+1}$ may involve changes in both the resources themselves and their economic circumstances.

This creates a decision problem.

The agent must determine what should happen to the resources it controls. It may retain them for future use, consume them, exchange them, transfer them, use them to produce other resources, or take actions intended to preserve them against loss.

These possibilities are not equivalent.

Exchanging one resource for another may change the form of wealth without necessarily increasing or reducing it. Using a resource in production may result in additional economic resources. Consuming a resource may remove it from the agent's wealth. A resource may also lose some or all of its economic significance without the agent choosing to do so.

The problem is therefore not simply how to acquire wealth.

It is how an agent should manage the wealth it already controls as that wealth changes through time.

This raises the central question:

> **How should an agent manage wealth through time?**

Answering this question requires first understanding the different ways wealth can change. In particular, we must distinguish changes in the resources themselves from changes in their economic valuation, and changes in the form of wealth from changes in the amount of wealth controlled by the agent.

Only then can we determine what it means for wealth to grow, what it means for wealth to be protected, and what other problems arise from managing wealth through time.

## The Transition of Wealth

If wealth is an economically realizable resource that an agent can control and retain through time, then wealth is necessarily temporal.

A resource controlled by an agent at one point in time may still be controlled by that agent later. But it may also be different in some respect. It may have changed form, been combined with another resource, generated another resource, been exchanged, transferred, consumed, impaired, or ceased to exist.

The most general representation of this is simply:

$$
w_t\rightarrow w_{t+1}.
$$

Here, $w_t$ represents a wealth resource at time $t$, and $w_{t+1}$ represents the corresponding resource at a later time.

The notation does not yet specify what happened between the two states. That is important. A transition from $w_t$ to $w_{t+1}$ does not, by itself, tell us whether wealth increased, decreased, or remained unchanged.

Consider a piece of land.

It may remain the same physical property:

$$
w_t=w_{t+1}.
$$

It may instead be exchanged for another resource:

$$
w_t\rightarrow w'_{t+1}.
$$

The agent may use the land to produce another resource while continuing to control the land:

$$
w_t\rightarrow\{w_t,w'_{t+1}\}.
$$

Or the land may be destroyed or otherwise cease to be an economically realizable resource:

$$
w_t\rightarrow\varnothing.
$$

These are different transitions, but they all belong to the same general problem: determining what happens to wealth through time.

The transition is also not determined solely by the wealth resource itself. It occurs through the interaction of the agent's actions with conditions outside the agent's control. A more complete representation is therefore:

$$
w_t\xrightarrow{a_t,e_t}w_{t+1},
$$

where $a_t$ represents the agent's action and $e_t$ represents the relevant external conditions.

The same resource can therefore undergo different transitions depending on what the agent does and what occurs around it.

But the resource itself is only one part of the transition.

The previous analysis of value established that value is positional:

$$
V(x\mid P)=U(F(P,a_x))-U(P).
$$

Consequently, the value of a wealth resource cannot necessarily be treated as a fixed property of the resource. If either the resource or the agent's position changes, its value may change as well.

For a wealth resource, the relevant transition is therefore:

$$
V(w_t\mid P_t)
\rightarrow
V(w_{t+1}\mid P_{t+1}).
$$

This produces a distinction between two different kinds of change.

The resource may change:

$$
w_t\neq w_{t+1},
$$

while its value to the agent may remain unchanged:

$$
V(w_t\mid P_t)=V(w_{t+1}\mid P_{t+1}).
$$

Conversely, the resource may remain substantially unchanged while its value changes:

$$
w_t=w_{t+1}
$$

while

$$
V(w_t\mid P_t)\neq V(w_{t+1}\mid P_{t+1}).
$$

Both may change simultaneously:

$$
w_t\neq w_{t+1}
$$

and

$$
V(w_t\mid P_t)\neq V(w_{t+1}\mid P_{t+1}).
$$

This distinction matters because a change in the resource is not necessarily a change in its value, and a change in value is not necessarily a change in the resource.

There is another distinction that follows directly from the definition of wealth.

Wealth requires both economic realizability and control. A transition can therefore affect not only the resource and its value, but also whether the resource remains wealth for the agent at all.

For example, an agent may transfer a resource to another agent:

$$
w_t^A\rightarrow w_{t+1}^B.
$$

The resource may continue to exist and may retain economic value, but it is no longer wealth controlled by agent $A$.

Likewise, a resource may remain physically intact while losing its economic realizability. Conversely, an economically realizable resource may become newly controlled by an agent:

$$
w\notin W_t^A
\rightarrow
w\in W_{t+1}^A.
$$

Thus, the transition of wealth is not simply a change in the market value of an object. It concerns the changing relationship between resources and the agent who controls them.

At this point, several distinct dimensions of transition have appeared:

$$
\boxed{
\begin{aligned}
\text{resource} &:\quad w_t\rightarrow w_{t+1}\\
\text{value} &:\quad V(w_t\mid P_t)\rightarrow V(w_{t+1}\mid P_{t+1})\\
\text{control} &:\quad C_t\rightarrow C_{t+1}\\
\text{economic realizability} &:\quad E_t\rightarrow E_{t+1}.
\end{aligned}
}
$$

These dimensions need not move together.

A resource can persist while its value changes. A resource can change form while preserving value. A resource can retain its economic value while leaving the agent's control. A resource can remain under the agent's control while becoming economically unrealizable.

This gives us a more precise starting point for the wealth problem.

The question is not initially whether wealth has “grown” or “declined.” Those are conclusions about a transition that require us to specify what changed and how the resulting state is being compared with the initial state.

The more fundamental object is the transition itself:

$$
\boxed{
(w_t,P_t,C_t,E_t)
\xrightarrow{a_t,e_t}
(w_{t+1},P_{t+1},C_{t+1},E_{t+1}).
}
$$

The problem of managing wealth therefore begins with understanding the possible transitions of wealth and the relationships among them.

Only after those transitions are characterized can we determine which transitions constitute accumulation, transformation, appreciation, depreciation, preservation, consumption, loss, or growth.

The next question is therefore not yet **how to grow wealth**, but:

> **What kinds of transitions can wealth undergo?**


## The Purpose of Wealth Management
The central problem of wealth management is deciding what to do with wealth that is already available.

An agent that possesses wealth has several possible courses of action. It can retain a resource, consume it, exchange it for another resource, transfer it, or use it in a way that produces additional resources.

These actions do not all have the same effect on wealth.

Consider an agent that possesses money. It can spend that money on consumption, exchange it for property, use it to purchase productive equipment, or invest it in an activity that produces a return.

In each case, the agent begins with one form of wealth and takes an action that changes what it possesses afterward.

The simplest case is consumption:

$$
\text{wealth}
\rightarrow
\text{consumption}.
$$

The wealth is exchanged for something that is consumed, and the original resource is no longer available afterward.

Another possibility is transformation:

$$
\text{wealth}_1
\rightarrow
\text{wealth}_2.
$$

The agent gives up one form of wealth and acquires another. Selling property and receiving money is an example. The form of the wealth changes, but this does not necessarily mean that wealth has been created or destroyed.

A third possibility is that wealth is used to produce additional wealth:

$$
\text{wealth}
\rightarrow
\text{production}
\rightarrow
\text{additional wealth}.
$$

A rental property may generate rent. Capital may generate a return. A business may produce a profit that can be retained as additional wealth.

Here the agent is not merely changing the form of its wealth. It is attempting to increase the wealth it controls.

There is also a different problem.

Wealth that already exists can be exposed to events that reduce or destroy it. Property can be damaged. Equipment can deteriorate. Financial claims can default. Resources can be lost or stolen.

An agent may therefore use some of its wealth to reduce the possibility or magnitude of such losses:

$$
\text{wealth}
\rightarrow
\text{protection}
\rightarrow
\text{preserved wealth}.
$$

We can therefore distinguish two fundamental objectives concerning the wealth itself.

The first is to **increase the wealth available to the agent**.

The second is to **preserve the wealth already available to the agent**.

We will refer to these as **growth** and **protection**.

$$
\boxed{
\begin{aligned}
\text{Growth} &: \quad \text{increase the wealth available to the agent}\\
\text{Protection} &: \quad \text{preserve the wealth available to the agent}
\end{aligned}
}
$$

This distinction is narrower than the broader question of how wealth affects the agent's position.

An agent may use wealth to acquire knowledge, improve its environment, increase its capabilities, or obtain other things that alter its position:

$$
\text{wealth}
\rightarrow
\text{other component of position}.
$$

Such an action may improve the agent's position even if it reduces the amount of wealth remaining.

Therefore, an increase in wealth and an improvement in position are not the same thing:

$$
\boxed{
\text{Growth of Wealth}
\neq
\text{Growth of Position}.
}
$$

The distinction is important because the Wealth Problem must first examine what happens to wealth itself before asking what the agent should ultimately do with that wealth.

The first question is therefore:

> **What does it mean for wealth itself to grow?**

That is the problem we examine next.

## Growing Wealth

The previous section distinguished the growth of wealth from the broader improvement of an agent's position.

The question is therefore:

> **What does it mean for wealth itself to grow?**

The simplest case is an increase in the amount of wealth controlled by an agent.

An agent possesses some wealth at time $t$ and, after some process has occurred, possesses more at time $t+1$:

$$
W_{t+1}>W_t.
$$

But this relationship requires some care.

Wealth can exist in many different forms. An agent may hold money, land, equipment, inventory, or financial claims. These resources cannot always be compared simply by counting them.

An agent may sell one property and acquire another. The number of properties has changed, but this does not necessarily mean that the agent's wealth has grown.

Likewise, an agent may exchange $\textdollar10,000$ of cash for equipment worth $\textdollar10,000$. The agent has less cash but now possesses another economic resource.

In both cases, wealth has changed in **form** without necessarily changing in **amount**.

We can therefore distinguish transformation from growth:

$$
\boxed{
\text{transformation}
\neq
\text{growth}.
}
$$

Growth requires something more than replacing one resource with another.

Consider instead a rental property that produces $\textdollar10,000$ in rent during a year. If the resulting income is retained as wealth, the agent has acquired an additional economic resource without giving up the original property:

$$
\text{property}
\rightarrow
\text{rent}
\rightarrow
\text{additional wealth}.
$$

The original resource remains while an additional resource has been produced.

A similar process occurs when capital is used in production:

$$
\text{capital}
\rightarrow
\text{production}
\rightarrow
\text{output}
\rightarrow
\text{wealth}.
$$

If the resulting output or proceeds are retained as economic resources, the agent can end the process with both the original productive resource and additional wealth.

This is different from simply exchanging one form of wealth for another.

The distinction can therefore be represented as:

$$
\begin{aligned}
\text{Transformation:}&\quad W_1\rightarrow W_2\\
\text{Growth:}&\quad W\rightarrow W+\Delta W,\qquad \Delta W>0.
\end{aligned}
$$

Growth is therefore associated with the creation or acquisition of additional economic resources rather than merely changing the form of resources already possessed.

But even this raises another problem.

Suppose an agent owns a property that increases in market price. Has the agent's wealth grown?

The agent may now be able to exchange the property for more money than before. But no new physical resource has necessarily been produced, and the agent may not have changed the property or its use.

The economic value associated with an existing resource can change even when the resource itself remains unchanged.

Similarly, a financial claim may become more valuable because the probability of repayment has increased, or because market conditions have changed.

This means that we must distinguish **creation of additional wealth** from **change in the value of existing wealth**.

The distinction can be represented as:

$$
\boxed{
\text{wealth creation}
\neq
\text{change in wealth valuation}.
}
$$

Both may increase the economic resources available to an agent, but they occur through different mechanisms.

This also shows why wealth growth cannot simply be identified with physical production.

An existing claim can appreciate without new physical production. An exchange can create a new economic claim. A productive activity can generate additional resources. Different processes can therefore produce an increase in the wealth controlled by an agent.

What they have in common is not necessarily the particular mechanism.

The common feature is that, after the process, the agent controls a greater quantity of economically realizable resources than before.

We can therefore state a provisional characterization:

$$
\boxed{
\text{Wealth growth occurs when an agent's stock of economically realizable resources increases.}
}
$$

This characterization still leaves an important question unresolved.

An increase in wealth does not tell us **why** the increase occurred or whether the process that produced it was beneficial to the agent's overall position.

An agent may acquire additional wealth by selling something it values highly, by accepting greater risk, or by sacrificing another component of its position.

Thus:

$$
\Delta W>0
$$

does not by itself imply:

$$
U(P_{t+1})>U(P_t).
$$

Wealth growth is therefore a property of the wealth itself, not a complete measure of the success of the agent's decisions.

This gives us an important distinction:

$$
\boxed{
\text{Growing wealth}
\neq
\text{improving position}.
}
$$

The former concerns what happens to the agent's economic resources.

The latter concerns what happens to the agent's position as a whole.

We have therefore identified one part of the Wealth Problem.

An agent can attempt to increase the wealth it controls through production, acquisition, exchange, investment, or other processes.

But increasing wealth is only useful if the resulting resources remain available.

That raises the complementary problem:

> **How can an agent protect the wealth it already possesses?**

That is the problem of protection.
