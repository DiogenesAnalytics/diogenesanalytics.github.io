## Introduction
Every agent seeks to improve its position.

An individual seeks greater security, wealth, knowledge, capability, and freedom. A corporation seeks capital, market position, technological capability, and strategic independence. A nation seeks resources, security, productive capacity, technological capability, and strategic autonomy.

These objectives differ, but they share a common structure. Each agent begins from some position and acts to transform that position into a more favorable one.

$$
P_{t+1}=F(P_t,a_t)
$$

Yet as position improves, something more fundamental changes.

The problem changes.

The problem facing an individual who cannot reliably obtain food is different from the problem facing one who has secure food and shelter. The problem facing a company struggling to survive is different from the problem facing one with abundant capital and an established market position. The problems facing a weak state are different from those facing a state with substantial productive, technological, financial, and strategic capabilities.

The agent does not simply solve the same problem repeatedly.

It crosses thresholds.

As one constraint becomes sufficiently small, another constraint becomes important. As one capability is acquired, new actions become possible. As new actions become possible, new objectives become attainable. The agent therefore moves through different regions of position space in which different problems dominate.

This suggests that the progression of position has a structure.

There is a ladder.

At the bottom of the ladder, an agent may be constrained by immediate survival. Higher up, the constraints may concern stability, security, independence, capability, control, or resilience. Higher still, the problem may cease to be the acquisition of some particular resource and become the management of an increasingly large space of possible futures.

The precise structure of this ladder is not obvious, nor should it be assumed in advance. The important observation is that the problem an agent must solve depends upon the position it occupies.

We can therefore think of the progression as

$$
P_0
\rightarrow
P_1
\rightarrow
P_2
\rightarrow
\cdots
$$

where each transition changes not only the position of the agent, but potentially the nature of the problem it faces.

This gives us a different way to think about strategic optimization.

The objective is not merely to make the current position better.

It is to reach a position from which a better class of positions becomes accessible.

In [*Strategic Solving and the Emergence of Inevitability*][1], we considered the other end of this process. As strategic optimization proceeds, the structure of reachable futures can contract until distinct actions no longer produce meaningfully distinct outcomes. A game can therefore become effectively solved while legal moves remain, because those remaining moves no longer materially affect the outcome.

The solved game describes a limiting condition.

But an agent does not begin at the limit.

It must travel there.

The interesting question is what lies between the starting position and inevitability.

What are the major thresholds? What problems dominate at each stage? How do resources change in value as position improves? And is there a recognizable trajectory through position space by which an agent progressively removes the constraints separating its current position from the solved game?

That trajectory is the subject of this article.

It is the **Golden Path of the Solved Game**.

## The Solved Game
A game is solved when strategic optimization has reduced the importance of the remaining choices to the point that the outcome is effectively determined.

The agent may still have actions available. It may still have decisions to make. But those decisions no longer produce materially different outcomes.

In [*Strategic Solving and the Emergence of Inevitability*][1], this process was described as a contraction of the set of materially distinct outcomes accessible from a position:

$$
\Delta(\Omega(P))\rightarrow0
$$

As this contraction occurs, the future becomes increasingly constrained. Actions that once produced meaningfully different trajectories cease to do so. The agent's position has become sufficiently strong that the remaining game can no longer substantially change its outcome.

This is the emergence of inevitability.

The solved game therefore provides a useful limiting condition for strategic optimization. It tells us what happens as an agent progressively removes the ability of the game to produce unfavorable outcomes.

But knowing the destination does not tell us how to get there.

An agent does not ordinarily move directly from its initial position to a solved position. It must pass through a succession of positions, each with its own constraints, resources, capabilities, and available actions.

The important question is therefore not what happens once the game is solved.

It is what happens **before** it is solved.

What changes as the agent approaches inevitability? What problems must be solved along the way? And why does solving one problem so often reveal another?

To answer those questions, we must examine the path through position itself.

## The Path Through Position
An agent moves toward the solved game through a sequence of positions.

$$
P_0\rightarrow P_1\rightarrow P_2\rightarrow\cdots\rightarrow P^*
$$

Each transition is produced by an action:

$$
P_{t+1}=F(P_t,a_t)
$$

At first glance, this appears to describe a simple process of improvement. The agent takes an action, its position improves, and it takes another action.

But the process is more interesting than this.

An improved position does not merely give the agent more of what it already had. It can change what the agent is capable of doing, what it is constrained by, and which outcomes are accessible.

A position with insufficient food creates one set of problems. A position with sufficient food but insecure access to that food creates another. Once food is secure, the constraint may shift elsewhere entirely.

The same pattern appears at larger scales.

A company struggling to remain solvent is primarily constrained by one set of problems. Once it has sufficient capital to survive, its constraints may shift toward production, distribution, competition, technology, or access to markets. A state that has secured its basic material resources may face very different constraints concerning productive capacity, technology, strategic independence, or control of its environment.

The position has changed, and therefore the problem has changed.

This means that the path through position cannot be understood simply as the accumulation of resources.

It is a succession of changing constraints.

At each position, some constraints limit the actions available to the agent more than others. Strategic improvement reduces those constraints and produces a new position in which a different set of constraints may become dominant.

The path can therefore be represented as

$$
P_t
\rightarrow
C_t
\rightarrow
a_t
\rightarrow
P_{t+1},
$$

where $C_t$ represents the constraints that matter at position $P_t$.

The crucial observation is that

$$
C_{t+1}\neq C_t
$$

may occur as the agent improves its position.

The agent is therefore not solving one problem over and over.

It is moving from one problem to another.

The question then becomes whether these changing problems occur in recognizable patterns—whether there are identifiable regions of position space in which particular classes of constraints tend to dominate.

If such regions exist, then the path toward the solved game may have a structure of its own.


## The Structure of the Ladder
If the dominant problem changes as position improves, then the path toward the solved game is not simply a continuous increase in some measure of position.

There are points at which the nature of the problem changes.

Consider an agent whose immediate access to food is uncertain. Improving its position may first require increasing its available food, securing a reliable source, or reducing the effort required to obtain it. Once that constraint is sufficiently reduced, additional food may provide little strategic improvement compared with solving another constraint.

The same pattern appears repeatedly.

A resource that is critical at one position may become relatively unimportant at another. An action that is necessary at one stage may become unnecessary later. A capability that was previously inaccessible may become available once another constraint has been removed.

The transition is therefore not simply

$$
P_t < P_{t+1}.
$$

It is a change in the structure of the problem being solved.

We can think of a positional regime as a region of position space in which a particular class of constraints dominates the agent's strategic decisions.

If $C(P)$ represents the dominant constraint at position $P$, then movement through position can produce transitions such as

$$
C(P_t)=C_1
\quad\rightarrow\quad
C(P_{t+1})=C_2.
$$

The boundary between these regimes is a threshold.

Crossing it does not necessarily require a discrete jump in the underlying position. Position may change continuously while the dominant problem changes abruptly. What matters is not that the position itself has discrete levels, but that different ranges of position can produce qualitatively different strategic problems.

This gives us a ladder.

The rungs of the ladder are not quantities of wealth, resources, or power. They are **classes of problems that become dominant at different positions**.

An agent can therefore move upward on the ladder without simply accumulating more of the same resource. It may instead exchange one constraint for another.

The progression might look like

$$
\text{Constraint}_1
\rightarrow
\text{Constraint}_2
\rightarrow
\text{Constraint}_3
\rightarrow
\cdots
$$

with each transition representing a change in what most limits the agent's ability to improve its position.

This also explains why the same action can have very different value at different positions.

An additional unit of a resource may produce a substantial improvement when that resource is the dominant constraint. Once the constraint has been sufficiently reduced, the same unit may produce little improvement, while another resource or capability becomes far more valuable.

The ladder is therefore not a hierarchy imposed upon the agent from outside.

It emerges from the changing relationship between the agent's position, its constraints, and its available actions.

The remaining question is what these recurring classes of problems actually are.

If the same structural progression appears across individuals, corporations, and states, then the ladder may reveal something more general about the path from an ordinary position toward the solved game.


## The Problems Along the Path
If the ladder emerges from changing constraints, then the next question is what those constraints actually look like.

The simplest case is survival.

An agent that cannot reliably maintain the resources required for continued existence is constrained by immediate replenishment. Food, water, shelter, and other essential resources determine what actions are possible. At this position, almost every other objective is subordinate to maintaining the position itself.

Once immediate survival is sufficiently secured, the problem changes.

The agent must make that security reliable.

A resource that is available today but may disappear tomorrow does not provide the same position as a resource that can be reliably replenished. The problem therefore shifts from acquisition to stability: establishing the conditions under which essential resources continue to be available.

From stability comes another problem.

What has been accumulated can be lost.

An agent with sufficient resources to survive but little ability to protect those resources remains vulnerable. Other agents, environmental changes, accidents, or failures can return the agent to a less favorable position.

The problem therefore becomes security.

Security changes the objective from simply maintaining resources to protecting the position those resources create.

Once security is sufficiently established, another constraint becomes visible.

Dependence.

An agent may possess substantial resources and still depend upon another agent, a particular supplier, a particular location, a particular technology, or a particular environmental condition. Its position may therefore be strong in one dimension while remaining constrained in another.

The problem becomes independence: increasing the range of actions that can be taken without relying upon a particular external condition.

At this point, the agent can begin solving a different class of problem.

It can build capability.

Capability is the ability to produce changes rather than merely respond to them. Knowledge, technology, capital, infrastructure, organization, and productive capacity allow an agent to create resources, alter constraints, and open new possible actions.

The progression can therefore be represented as

$$
\text{Survival}
\rightarrow
\text{Stability}
\rightarrow
\text{Security}
\rightarrow
\text{Independence}
\rightarrow
\text{Capability}.
$$

These should not be understood as universal discrete levels through which every agent must pass in exactly this order. They describe recurring changes in the dominant problem as an agent's position improves.

The important transition is between the problems themselves.

Survival asks whether the agent can continue.

Stability asks whether that continuation can be made reliable.

Security asks whether the resulting position can be protected.

Independence asks whether the position can be maintained without excessive dependence on external conditions.

Capability asks what the agent can produce or change from that position.

Each solution changes the space of available actions.

And once an agent can substantially change its environment, the problem changes again.

The question is no longer merely whether the agent can act.

It becomes how much of the surrounding environment it can control, how robustly it can preserve its position, and how far it can constrain the set of unfavorable futures available to the game.

The ladder therefore continues beyond capability.

Those later rungs are where the path begins to converge with the solved game itself.
