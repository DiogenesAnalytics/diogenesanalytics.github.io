## Introduction
Water is essential for life. Every human civilization has depended upon it, every society has developed strategies to acquire and manage it, and countless conflicts have been shaped by its availability and control. Despite its obvious importance, discussions of water typically focus on its biological, environmental, or economic roles.

This article asks a different question:

> **What role does water play in an agent's position within the game of life?**

Within the Position framework, an agent's position is the collection of information that determines the threats it faces, the opportunities it can exploit, and the future states it can potentially reach. Position is therefore not merely a description of an agent's current circumstances; it is the informational state from which all future decisions and outcomes emerge.

The central hypothesis of this article is that **water constitutes a fundamental dimension of position**. There exist classes of threats and opportunities whose outcomes cannot be determined without information about an agent's relationship to water. Consequently, any sufficiently complete description of an agent's position must include a water dimension.

The purpose of this article is not to study water itself. Rather, it is to investigate why water is a fundamental dimension of position, identify the kinds of information that comprise an agent's water position, and explore how changes in that position expand or constrain the set of possible future states.

More broadly, this article serves as a case study in the methodology of identifying and analyzing the dimensions of position. If water can be understood as a dimension of position, then the same methodology may be applied to other fundamental dimensions of the game of life, including food, energy, health, knowledge, and wealth.
