## Introduction
Wealth is one of the most consequential concepts in human society. Individuals accumulate it, exchange it, preserve it, invest it, consume it, and pass it between generations. Entire institutions, markets, and economic systems have developed around its creation and management. Yet despite its central role in human life, wealth is often treated as though its meaning were self-evident.

We commonly describe wealth in terms of money, income, assets, or net worth. These measures are useful, but none of them is wealth itself. A person may hold a large quantity of cash, own valuable land, control a productive business, or possess claims on future income. These forms of wealth differ substantially in their liquidity, risk, durability, and ability to provide access to resources, yet they can all represent wealth.

This raises a more fundamental question:

> **What is wealth?**

The question is more difficult than it initially appears because wealth is inherently related to time. An agent does not merely possess resources; an agent may possess the ability to command resources in the future. A resource that can be consumed today, a claim that can be exchanged tomorrow, and an asset that can generate resources for decades are different manifestations of this same underlying capacity.

This temporal character of wealth creates a fundamental problem. Wealth that has been generated must somehow retain its capacity to command resources until that capacity is needed. Holding money, for example, preserves a nominal quantity of monetary units, but does not guarantee preservation of their purchasing power. Other forms of wealth may fluctuate substantially in market price while continuing to represent claims on scarce resources or future production. Consequently, understanding wealth requires distinguishing the underlying capacity represented by an asset from the price at which that asset happens to trade.

The problem therefore extends beyond the accumulation of money or assets. It concerns the acquisition, preservation, and eventual deployment of an agent's capacity to command resources across time.

This article develops a definition of wealth from that perspective. It examines the relationship between wealth and resources, the role of time in determining the persistence of wealth, and the distinction between value and market price. It then considers the different forms through which wealth may be represented and the fundamental problems involved in generating and storing it.

The objective is not to develop a theory of investing or portfolio management. Those are problems that arise after the more fundamental concept has been established. The objective here is to identify what wealth actually is, so that the subsequent problems of financial position, capital staging, and wealth management can be formulated on a coherent foundation.

The central proposition developed in this article is:

> **Wealth is an agent's capacity to command resources across time.**


## Wealth and Resources
Wealth is not the same thing as a resource. A resource is something from which an agent can obtain some useful outcome. Food can provide energy, land can provide space, water can sustain life, and a machine can produce goods. Resources therefore have value because they enable an agent to satisfy needs, pursue objectives, or reach future states.

Wealth, however, need not consist of the resources themselves. An agent can possess wealth without directly possessing the resource it will eventually consume. Money, for example, can be exchanged for food without being food itself. A financial claim can provide access to future income without containing that income. Ownership of productive capital can provide access to future production without the agent directly possessing the goods that will be produced.

What these different forms have in common is that they provide an agent with some capacity to command resources.

This suggests a distinction between a resource and a claim upon resources. The resource is the thing that can ultimately be used. The claim is the mechanism through which an agent obtains access to it.

An agent holding a quantity of food possesses the resource directly. An agent holding money possesses a claim that can potentially be exchanged for food. An agent owning a productive business possesses a claim upon a stream of future production. These situations are materially different, but each gives the agent some ability to command resources.

We can therefore distinguish the immediate resource from the capacity to obtain resources:

$$
\text{Resource} \neq \text{Capacity to command resources}.
$$

The distinction becomes particularly important when considering resources that are not immediately available. An agent may possess a claim to future wages, future production, rental income, or the proceeds from an asset sale. These claims can have value even though the resources they ultimately command do not yet exist in the agent's possession.

Wealth therefore concerns not only what an agent currently possesses, but also what the agent can command as a consequence of what it possesses.

This provides a more general interpretation of wealth. Rather than identifying wealth with money, assets, or any particular class of objects, we can treat those things as different mechanisms through which an agent acquires the capacity to command resources.

Under this interpretation, the relevant question is not simply how much of some asset an agent possesses, but what that possession enables the agent to obtain.

A quantity of cash, a parcel of land, an ownership stake in a company, and a claim to future income may have entirely different physical and financial characteristics. Yet each can constitute wealth because each can provide an agent with access to resources.

The concept of wealth therefore begins to emerge as a property of the relationship between an agent and the resources available to it, rather than as a property of any particular object.

This distinction will become more important when time is introduced. An agent's capacity to command resources may extend into the future, and the ability to preserve that capacity is itself a fundamental property of wealth. Thus, to understand wealth completely, it is necessary to consider not only what resources an agent can command, but **when** those resources can be commanded.

## Wealth Across Time
The distinction between resources and the capacity to command resources becomes more significant when time is introduced. An agent may possess a resource today, possess a claim to a resource tomorrow, or possess an asset that can generate resources continuously over a future period. These are different temporal configurations of the same underlying capacity.

Consider an agent who possesses enough food for one day. The agent can command a quantity of resources immediately, but that capacity does not extend indefinitely into the future. Another agent may possess the same quantity of food today together with a reserve sufficient for the following year. The two agents have similar present resources, but they do not possess the same capacity to command resources across time.

This distinction can be represented by treating resource access as a function of time. Let

$$
R(t)
$$

represent the resources an agent can command at time $t$. Wealth is not therefore adequately described by the value of $R(t)$ at a single instant. It concerns the structure of the agent's capacity to command resources over an interval of time.

An agent may increase this capacity in several ways. It may store resources directly, acquire claims on resources that already exist, acquire claims on future production, or acquire productive assets capable of generating additional resources. These mechanisms differ substantially, but they all alter the agent's future resource access.

For example, cash held today can be exchanged for resources in the future. A bond can provide contractual payments at specified future dates. A productive business can generate a stream of future income. Land can provide continuing access to a scarce physical resource. None of these assets is identical to the resources the agent will ultimately consume, yet each can provide a capacity to command those resources at future times.

This suggests that wealth is inherently intertemporal. An asset constitutes wealth not merely because it can command resources now, but because it provides some capacity to command resources at one or more points in the future.

We can therefore represent an agent's wealth more generally as a set of future resource claims:

$$
W = {C(t_1), C(t_2), \ldots, C(t_n)},
$$

where $C(t_i)$ represents the agent's capacity to command resources at time $t_i$.

This representation immediately reveals why two agents possessing the same nominal amount of wealth can nevertheless occupy different financial states. Suppose one agent possesses $\textdollar100,000$ that can be deployed immediately, while another possesses $100,000 that cannot be accessed for twenty years. Their nominal wealth may be identical:

$$
W_A = W_B = $100{,}000
$$

but their capacities to command resources over time are not:

$$
C_A(t) \neq C_B(t).
$$

The difference is not merely one of convenience. The timing of resource access determines which actions are feasible. An agent that cannot command resources until some future date cannot use those resources to respond to an opportunity or threat that occurs before that date.

Time therefore introduces another property of wealth: **availability**.

A claim on resources is more useful for a particular purpose when those resources are available at the time they are required. This means that the value of wealth cannot be separated entirely from the timing of the resources it can command.

The problem becomes even more significant when the quantity of resources that can be commanded changes over time. An agent holding a fixed quantity of money does not necessarily possess a fixed quantity of real purchasing power. If the prices of resources increase, the same nominal amount of money commands fewer resources.

Let $M$ represent a quantity of money and let $P_B(t)$ represent the price of a reference basket of resources at time $t$. The purchasing power of that money relative to the basket can be represented as

$$
Q_M(t) = \frac{M}{P_B(t)}.
$$

If $M$ remains constant while $P_B(t)$ increases, then

$$
\frac{dQ_M}{dt} < 0.
$$

The nominal quantity of wealth has remained unchanged, but the quantity of resources that it can command has declined.

This distinction is fundamental. Preserving the numerical quantity of an asset is not necessarily equivalent to preserving wealth. What must be preserved is the agent's capacity to command resources.

The same reasoning applies in the opposite direction. An asset may increase substantially in market price while the underlying capacity it provides to command resources changes little. Conversely, an asset's market price may decline while its underlying resource-producing capacity remains substantial.

Thus, the temporal problem of wealth cannot be reduced to maintaining a nominal balance. It is the problem of maintaining, and potentially increasing, the capacity to command resources across time.

This observation leads directly to a deeper question. If wealth is a capacity to command resources across time, **how should that capacity be measured?** In particular, what is the relationship between the value represented by an asset and the price at which that asset can be exchanged?

That distinction between **value and price** is central to understanding wealth.


## Value and Price
If wealth is an agent's capacity to command resources across time, then the value of an asset cannot be determined solely by observing the quantity of money for which it can currently be exchanged. Market price provides an important measurement, but it measures an exchange relation at a particular point in time. Wealth concerns the underlying capacity that the asset provides to command resources.

Let $P_a(t)$ denote the market price of an asset $a$ at time $t$. This is the amount of money required to acquire the asset under prevailing market conditions. If the asset is sold at that time, $P_a(t)$ also represents the quantity of money that can be obtained in exchange for it.

But money itself is a claim on resources, and the quantity of resources that a given amount of money can command changes over time. Let $P_B(t)$ denote the price of a reference basket of resources. The purchasing power of asset $a$, measured in units of that basket, can then be represented as

$$
R_a^B(t) = \frac{P_a(t)}{P_B(t)}
$$

This quantity answers a different question from the market price. Rather than asking how many monetary units the asset commands, it asks how much of the reference resource basket those monetary units can command.

For cash, this distinction is immediate. Suppose an agent holds a fixed quantity of money $M$. Its nominal value remains

$$
P_M(t) = M
$$

But its purchasing power is

$$
R_M^B(t) = \frac{M}{P_B(t)}
$$

If the price of the resource basket increases while $M$ remains constant, then

$$
\frac{dR_M^B}{dt}<0
$$

The agent has preserved the nominal quantity of money but lost part of its capacity to command resources. Nominal preservation and preservation of wealth are therefore not equivalent.

This also demonstrates why inflation is fundamentally a wealth-storage problem. An agent that has already generated wealth must preserve its capacity to command resources until that wealth is eventually deployed. If the representation in which wealth is stored loses purchasing power, then some of the wealth has effectively been lost even though the nominal balance has not changed.

Market price introduces a different problem. The market price of an asset can change even when the underlying resource or productive capacity represented by the asset changes very little.

Consider land. Let $Q_L$ represent the physical quantity of a particular parcel of land and let $P_L(t)$ represent its market price. The market price may change substantially:

$$
P_L(t_1) \neq P_L(t_2)
$$

while the physical quantity of land remains essentially unchanged:

$$
Q_L(t_1) = Q_L(t_2)
$$

A decline in the market price therefore does not imply that the quantity of land controlled by the owner has declined by the same proportion. The market may have revised its valuation of the land without altering the underlying physical resource.

The same distinction appears in ownership of productive enterprises. Suppose an agent owns $n$ shares of a company with $N$ shares outstanding. The agent's ownership fraction is

$$
S=\frac{n}{N}
$$

The market price of those shares may fluctuate substantially while $S$ remains constant. The market value of the claim has changed, but the agent's proportional ownership of the underlying enterprise has not necessarily changed.

These examples reveal that at least three quantities must be distinguished:

$$
\boxed{
\text{Market Price}
\neq
\text{Purchasing Power}
\neq
\text{Underlying Resource Claim}
}
$$

Market price describes an exchange relation between an asset and money. Purchasing power describes the quantity of resources that the asset can command through exchange. An underlying resource claim describes the resources, productive capacity, or other economic capabilities to which ownership of the asset provides access.

These quantities may move together, but they need not.

An asset can increase in market price without providing a corresponding increase in underlying resources. An asset can decrease in market price while continuing to provide substantial access to resources. And an asset can maintain a constant nominal price while its purchasing power declines because the prices of resources increase.

This means that market price is best understood as an **observable market measurement of an asset**, rather than as a complete definition of its value.

The deeper quantity of interest is the capacity that ownership of the asset gives an agent to command resources across time. Let

$$
V_a(t)
$$

denote this capacity for asset $a$. Unlike market price, $V_a(t)$ need not be directly observable. It depends upon the resources the asset represents, the timing of those resources, the agent's ability to access them, and the uncertainty surrounding that access.

We therefore arrive at an important distinction:

$$
P_a(t) = \text{what the market currently exchanges the asset for}
$$

while

$$
V_a(t) = \text{what capacity to command resources the asset provides}
$$

The two quantities are related, but they are not identical.

This distinction also explains why the concept of wealth cannot be reduced to a single market valuation. A market price is useful because it provides a common unit for comparing heterogeneous assets. But the underlying wealth represented by those assets may consist of fundamentally different forms of resource access: immediate purchasing power, ownership of scarce physical resources, claims on future income, productive capacity, or other forms of future command over resources.

The problem of measuring wealth is therefore more complicated than measuring market capitalization or adding together account balances. To understand an agent's wealth, we must understand not only what its assets could be exchanged for today, but what resources those assets enable the agent to command, when those resources become available, and how reliably that capacity persists through time.

This distinction between price and value provides the foundation for examining the different forms through which wealth can be represented.
