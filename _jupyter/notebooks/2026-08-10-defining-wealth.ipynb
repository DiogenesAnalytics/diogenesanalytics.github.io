## Introduction
Wealth is one of the most consequential concepts in human society. Individuals accumulate it, exchange it, preserve it, invest it, consume it, and pass it between generations. Entire institutions, markets, and economic systems have developed around its creation and management. Yet despite its central role in human life, wealth is often treated as though its meaning were self-evident.

We commonly describe wealth in terms of money, income, assets, or net worth. These measures are useful, but none of them is wealth itself. A person may hold a large quantity of cash, own valuable land, control a productive business, or possess claims on future income. These forms of wealth differ substantially in their liquidity, risk, durability, and ability to provide access to resources, yet they can all represent wealth.

This raises a more fundamental question:

> **What is wealth?**

The question is more difficult than it initially appears because wealth is inherently related to time. An agent does not merely possess resources; an agent may possess the ability to command resources in the future. A resource that can be consumed today, a claim that can be exchanged tomorrow, and an asset that can generate resources for decades are different manifestations of this same underlying capacity.

This temporal character of wealth creates a fundamental problem. Wealth that has been generated must somehow retain its capacity to command resources until that capacity is needed. Holding money, for example, preserves a nominal quantity of monetary units, but does not guarantee preservation of their purchasing power. Other forms of wealth may fluctuate substantially in market price while continuing to represent claims on scarce resources or future production. Consequently, understanding wealth requires distinguishing the underlying capacity represented by an asset from the price at which that asset happens to trade.

The problem therefore extends beyond the accumulation of money or assets. It concerns the acquisition, preservation, and eventual deployment of an agent's capacity to command resources across time.

This article develops a definition of wealth from that perspective. It examines the relationship between wealth and resources, the role of time in determining the persistence of wealth, and the distinction between value and market price. It then considers the different forms through which wealth may be represented and the fundamental problems involved in generating and storing it.

The objective is not to develop a theory of investing or portfolio management. Those are problems that arise after the more fundamental concept has been established. The objective here is to identify what wealth actually is, so that the subsequent problems of financial position, capital staging, and wealth management can be formulated on a coherent foundation.

The central proposition developed in this article is:

> **Wealth is an agent's capacity to command resources across time.**
