## Introduction
In previous work, we introduced the concept of **position** as the state of an agent within a game. Rather than treating position as a single quantity, we argued that it is more naturally represented as a vector,

$$
P = (x_1, x_2, \ldots, x_n),
$$

where each coordinate represents a fundamental dimension that influences the agent's ability to achieve future outcomes.

This representation reflects an important observation: position is inherently multidimensional. An agent may possess abundant financial resources but little knowledge. Another may be highly skilled yet physically constrained. A third may enjoy excellent health but lack the social relationships necessary to accomplish certain goals. No single scalar quantity can adequately describe these differences because each dimension enables access to a distinct class of attainable positions.

This naturally raises an important theoretical question:

> **Which variables deserve to be included in the position vector?**

Not every variable that influences outcomes qualifies as a fundamental dimension of position. The weather may influence the outcome of a baseball game, and the stock market may influence an investor's wealth, but neither necessarily constitutes an independent coordinate of the agent's position. A rigorous theory therefore requires a criterion for distinguishing **fundamental dimensions** from variables that merely influence outcomes.

One candidate that has received surprisingly little formal attention is **environment**.

At first glance, environment appears to be little more than external context. Unlike wealth, knowledge, or health, it does not appear to be a property of the agent at all. Instead, it seems to describe the world surrounding the agent. Consequently, it is tempting to treat environment as something outside the Position framework rather than as one of its dimensions.

Yet this intuition quickly encounters a difficulty.

Consider two agents who are identical in every measurable respect. They possess the same knowledge, the same wealth, the same health, the same skills, and the same preferences. Now place these agents into different environments. Despite being identical as individuals, they immediately acquire different sets of opportunities, constraints, and possible futures. One may be able to start a business, while the other cannot. One may enjoy political stability, while the other faces civil unrest. One may have access to abundant natural resources, while the other does not.

The difference is not the agent.

The difference is the environment.

This observation motivates the central question of this article:

> **Is environment a fundamental dimension of position?**

More precisely,

> **Can there exist attainable positions that cannot be reached solely through improving the agent, but instead require changing the environment in which the agent operates?**

This article argues that the answer is yes.

We will show that environment satisfies the same criterion used to identify other fundamental dimensions of position. Different environments alter the set of feasible actions available to an agent, modify the costs associated with those actions, and ultimately determine which future positions are reachable. Environment is therefore not merely background context; it is an independent coordinate in the position vector.

Establishing this result extends the Position framework in an important way. It implies that improving position is not limited to improving the agent. Rational agents must also consider improving their relationship to the environments in which they operate. This observation provides the theoretical foundation for later work on environmental access, mobility, jurisdictional strategy, visa optimization, and the construction of an environment portfolio.

The objective of this article is therefore not to develop a theory of relocation or immigration. Rather, it is to establish a more fundamental result:

> **Environment is a fundamental dimension of position because it determines classes of attainable positions that cannot be reached solely through changes to the agent.**

## Why Environment Matters
Every agent exists within an environment. Whether that environment is a rainforest, a city, a nation, or a digital ecosystem, it provides the conditions under which the agent survives, acts, and competes. The environment determines what resources are available, what hazards are present, what actions are possible, and what opportunities may be pursued.

It is tempting to treat environment as merely the backdrop against which life unfolds. After all, environments exist independently of the agent. Unlike knowledge, health, or financial resources, they are not possessed by the individual. Consequently, environment is often viewed as external context rather than as a component of position.

This intuition, however, overlooks a fundamental property of environments.

An agent's environment does not merely influence the outcome of actions. It determines which actions are available in the first place. The same agent, possessing identical knowledge, wealth, health, skills, and preferences, may have access to entirely different futures simply because they occupy different environments.

Consider two otherwise identical individuals.

One lives in a politically stable country with strong property rights, reliable infrastructure, and unrestricted access to global markets. The other lives in a region experiencing civil conflict, limited infrastructure, and severe restrictions on economic activity.

Although the agents themselves are identical, their positions are clearly not.

The first agent can pursue opportunities that are simply unavailable to the second. Likewise, the second agent faces risks and constraints that do not exist for the first. The difference lies not within the agents themselves, but within the environments they inhabit.

This observation suggests that environment influences something more fundamental than the outcome of individual decisions.

It influences the set of positions that are attainable.

An agent's future is determined not only by who they are, but also by where they are. Certain positions can only be reached from particular environments, while others remain inaccessible regardless of the agent's internal capabilities.

If position is intended to describe an agent's ability to navigate the game they are playing, then any variable that systematically changes the set of attainable future positions deserves careful consideration.

The remainder of this article argues that environment satisfies this criterion. Rather than serving as passive background context, environment functions as an independent dimension of position because it fundamentally alters the landscape of opportunities, constraints, and reachable futures available to the agent.
