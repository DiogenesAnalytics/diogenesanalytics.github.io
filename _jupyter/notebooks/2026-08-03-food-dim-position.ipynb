## Introduction
Finance has played a central role in the development of human civilization. The ability to store value, exchange resources, allocate capital, and transfer purchasing power across time has shaped the organization of societies, the development of institutions, and the expansion of human capability. Throughout history, individuals and societies have developed increasingly sophisticated systems for acquiring, preserving, and deploying financial resources. Despite its obvious importance, discussions of finance typically focus on money, markets, investment returns, or economic activity.

This article asks a different question:

> **What role does finance play in an agent's position within the game of life?**

Within the Position framework, an agent's position is the collection of information that determines the threats it faces, the opportunities it can exploit, and the future states it can potentially reach. Position is therefore not merely a description of an agent's current circumstances; it is the informational state from which all future decisions and outcomes emerge.

The central hypothesis of this article is that **finance constitutes a fundamental dimension of position**. There exist classes of threats and opportunities whose outcomes cannot be determined without information about an agent's relationship to financial resources. Consequently, any sufficiently complete description of an agent's position must include a finance dimension.

Finance position is not simply a measurement of wealth. Wealth represents only one aspect of an agent's financial state. The finance dimension describes the broader relationship between an agent and the resources required to pursue future states. It includes the ability to acquire resources, preserve them through time, exchange them with others, allocate them toward productive uses, and maintain flexibility in the face of uncertainty.

Two agents may possess similar present resources yet occupy fundamentally different financial positions. One may have stable income, low obligations, high liquidity, and broad access to opportunities. Another may possess equivalent assets but face significant liabilities, limited flexibility, and vulnerability to disruption. Their current circumstances may appear similar, but the set of future positions available to each agent is different.

This observation motivates the central question of this article:

> **What information is required to describe an agent's financial position, and why does that information affect the set of attainable future states?**

A rigorous theory of position requires distinguishing fundamental dimensions from variables that merely influence outcomes. Financial markets, asset prices, and individual transactions may affect an agent's circumstances, but the deeper question is whether finance represents an independent coordinate of position — a dimension whose variation creates differences in attainable futures that cannot be explained by other dimensions alone.

This article argues that finance satisfies this criterion. An agent's financial position determines access to resources, the ability to withstand adverse events, the capacity to make commitments, and the range of opportunities available. Differences in financial position create differences in the future states that agents can reach, even when other dimensions of position remain identical.

The purpose of this article is therefore not to develop a theory of investing, markets, or personal finance. Rather, it is to investigate why finance is a fundamental dimension of position, identify the information that comprises an agent's financial position, and explore how changes in that position expand or constrain the space of possible futures.

More broadly, this article serves as another case study in the methodology of identifying and analyzing the dimensions of position. If finance can be understood as a dimension of position, then the same framework can be applied to other fundamental dimensions of the human game, revealing the structure of the position space within which agents make decisions and pursue future outcomes.
