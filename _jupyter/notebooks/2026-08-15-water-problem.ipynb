## The Water Problem
Water is among the most basic requirements of human life. Every human being must continually obtain water, yet access to water is neither uniform nor guaranteed. Water may be abundant or scarce, clean or contaminated, nearby or distant, reliable or seasonal, inexpensive or extraordinarily costly to obtain.

It is therefore tempting to describe the water problem simply as a problem of scarcity: humans need water, and sometimes there is not enough. But this description is incomplete. An abundance of water in a distant location may be of little practical value. A nearby source may be unusable because it is contaminated. A plentiful source may be available today but unreliable tomorrow. Water may even be physically abundant while remaining inaccessible because the agent lacks the means or authority to obtain it.

The water problem is consequently more fundamental than the question of how much water exists. It is the problem of **maintaining sufficient access to usable water over time**.

This problem confronts human beings at every scale. An individual must obtain enough water to survive. A household must secure a reliable supply. A farmer must obtain enough water to sustain crops. A city must develop systems capable of supplying an entire population. A civilization must construct institutions and infrastructure capable of managing water across space and time.

Human history can therefore be viewed, in part, as a history of attempts to solve water problems. Wells, irrigation systems, reservoirs, aqueducts, canals, pipelines, water treatment, and desalination are all different means by which human beings have altered their relationship to the constraints imposed by water.

The purpose of this article is to understand this problem in its most general form. What exactly makes water a problem for an agent? What constraints does it impose? What determines whether water is available and usable? What opportunities arise when those constraints are overcome? And what information must an agent possess in order to make effective decisions concerning water?

Only after these questions have been examined can we ask the further theoretical question of whether the information revealed by the water problem constitutes a distinct **dimension of position** within the game of life.
