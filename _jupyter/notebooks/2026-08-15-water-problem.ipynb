## The Water Problem
Water is among the most basic requirements of human life. Every human being must continually obtain water, yet access to water is neither uniform nor guaranteed. Water may be abundant or scarce, clean or contaminated, nearby or distant, reliable or seasonal, inexpensive or extraordinarily costly to obtain.

It is therefore tempting to describe the water problem simply as a problem of scarcity: humans need water, and sometimes there is not enough. But this description is incomplete. An abundance of water in a distant location may be of little practical value. A nearby source may be unusable because it is contaminated. A plentiful source may be available today but unreliable tomorrow. Water may even be physically abundant while remaining inaccessible because the agent lacks the means or authority to obtain it.

The water problem is consequently more fundamental than the question of how much water exists. It is the problem of **maintaining sufficient access to usable water over time**.

This problem confronts human beings at every scale. An individual must obtain enough water to survive. A household must secure a reliable supply. A farmer must obtain enough water to sustain crops. A city must develop systems capable of supplying an entire population. A civilization must construct institutions and infrastructure capable of managing water across space and time.

Human history can therefore be viewed, in part, as a history of attempts to solve water problems. Wells, irrigation systems, reservoirs, aqueducts, canals, pipelines, water treatment, and desalination are all different means by which human beings have altered their relationship to the constraints imposed by water.

The purpose of this article is to understand this problem in its most general form. What exactly makes water a problem for an agent? What constraints does it impose? What determines whether water is available and usable? What opportunities arise when those constraints are overcome? And what information must an agent possess in order to make effective decisions concerning water?

Only after these questions have been examined can we ask the further theoretical question of whether the information revealed by the water problem constitutes a distinct **dimension of position** within the game of life.

## The Existence of the Water Problem
The most basic fact of the water problem is that water is a **continuous requirement**. An agent cannot acquire water once and then permanently solve the problem. Water is consumed, lost, or rendered unusable, and therefore access must be maintained over time.

At the same time, water is not distributed according to the needs of the agents that depend upon it. Water exists in particular quantities, locations, forms, and times. Rainfall is seasonal. Rivers change their flow. Groundwater may be difficult to access. Sources may become contaminated. A population may grow faster than its water infrastructure. Thus, the existence of water somewhere in the environment does not guarantee that an agent can obtain the water it requires.

This creates a fundamental distinction between **water that exists** and **water that is available to the agent**.

We can express the simplest version of the problem as:

$$
W_{\text{required}}(t) \leq W_{\text{accessible}}(t)
$$

For an agent to satisfy its water requirements at time (t), the amount of usable water it can actually access must be at least as great as the amount it requires.

The inequality is deceptively simple. The difficulty lies in determining what counts as *accessible* water.

A body of water may be physically present but inaccessible because it is too distant, too deep, or beyond the agent's technological capabilities. Water may be accessible but unusable because it is contaminated. A source may provide sufficient water under normal conditions but fail during a drought. Water may be available in sufficient quantity but only at a cost greater than the agent can bear.

Consequently, the water problem cannot be reduced to the question:

> **Is there water?**

The relevant question is:

> **Can the agent obtain sufficient usable water, where and when it is needed, at an acceptable cost, and with sufficient reliability to support its activities?**

This problem exists regardless of whether an agent is an isolated individual drawing water from a stream, a farmer irrigating a field, a city operating a municipal water system, or a civilization managing water across an entire region. The mechanisms differ dramatically with scale, but the underlying constraint remains: **the agent must maintain access to sufficient usable water in order to sustain its activities through time.**

This gives us the basic form of the water problem. We can now examine the different ways in which that problem can arise—through insufficient quantity, lack of access, poor quality, unreliable supply, spatial or temporal mismatch, and excessive cost—and the different means by which agents attempt to overcome those constraints.

## The Basic Water Constraint
The simplest form of the water problem is a comparison between what an agent requires and what it can actually access. Let $W_{\text{required}}(t)$ represent the amount of usable water an agent requires at time $t$, and let $W_{\text{accessible}}(t)$ represent the amount of usable water the agent can obtain at that time. The basic constraint is therefore:

$$
W_{\text{required}}(t) \leq W_{\text{accessible}}(t)
$$

When this condition is satisfied, the agent has sufficient water to meet the requirement under consideration. When it is not satisfied, a water deficit exists:

$$
W_{\text{deficit}}(t)
=
W_{\text{required}}(t)
-
W_{\text{accessible}}(t)
>
0
$$

This formulation is intentionally simple. It establishes the fundamental structure of the problem without yet attempting to account for all of the factors that determine whether water is accessible.

The distinction between **required** and **accessible** water is important. An agent may require a certain quantity of water, while the environment may contain a much greater quantity. The existence of water in the environment does not, by itself, satisfy the agent's requirement. What matters is the quantity that the agent can actually obtain and use.

The problem therefore concerns the relationship between an agent and its water supply, rather than the absolute quantity of water present in the environment.

Furthermore, water requirements and water access are not necessarily static. An agent consumes water over time, while its sources of water may vary in quantity or accessibility. A more useful representation therefore considers the agent's water stock.

Let $S(t)$ represent the agent's stock of usable water at time $t$. Let $I(t)$ represent water acquired during the interval, and let $C(t)$ represent water consumed or otherwise lost. The stock then evolves according to:

$$
S(t+1)=S(t)+I(t)-C(t)
$$

The agent must maintain sufficient stock to satisfy its requirements:

$$
S(t)\geq S_{\min}(t)
$$

where $S_{\min}(t)$ represents the minimum usable water stock required to avoid whatever consequence is being considered.

This immediately introduces the temporal character of the water problem. An agent may possess sufficient water at one moment while nevertheless facing a water problem in the future. Conversely, an agent may temporarily lack water while possessing a reliable means of obtaining more before the shortage becomes consequential.

Thus, the relevant question is not simply:

> **How much water does the agent have?**

It is:

> **Can the agent maintain sufficient access to usable water over the period relevant to its objectives?**

This distinction will become increasingly important as the problem is examined in greater detail. The simple water balance does not yet tell us *why* water is accessible or inaccessible, *whether* it is usable, *how reliable* the supply is, or *what it costs* to obtain it. Those questions constitute the different forms of the water problem examined next.


## The Forms of the Water Problem
The basic water constraint tells us that an agent must maintain access to sufficient usable water over time. But it does not tell us why that condition might fail.

An agent can face a water problem even when a large quantity of water exists in its environment. The problem may arise because there is not enough water, because the available water cannot be reached, because it cannot safely be used, because its supply is unreliable, because it is available in the wrong place or at the wrong time, or because obtaining it requires resources that the agent cannot afford.

The most direct form of the problem is therefore a shortage of water. An agent may simply be unable to obtain enough water because the relevant supply is insufficient. A drought may reduce the flow of a river. A well may produce too little water. A reservoir may fall below the level required to supply a population. In this case, the basic constraint fails because the quantity of water accessible to the agent is less than the quantity it requires:

$$
W_{\text{accessible}}(t)<W_{\text{required}}(t)
$$

This is the form of the problem most naturally described as **scarcity**. But scarcity is only one way in which the constraint can fail.

Water may exist in sufficient quantity without being accessible to the agent. A river several hundred miles away may contain enormous quantities of freshwater, but that water does not satisfy the requirements of an agent that has no means of transporting it. Groundwater may exist beneath a settlement but remain inaccessible because the agent lacks the technology or energy required to extract it.

This creates an important distinction between the quantity of water physically present and the quantity the agent can actually obtain. If $W_{\text{physical}}(t)$ represents water physically present in the relevant environment, then:

$$
W_{\text{accessible}}(t)\leq W_{\text{physical}}(t)
$$

The difference between these quantities reflects, among other things, geography, technology, infrastructure, and control. The development of wells, canals, pipelines, and pumps can therefore be understood as an attempt to transform physically existing water into water that is accessible to the agent.

Even water that can be obtained may not be usable. A source may contain pathogens, toxic chemicals, excessive salinity, or other contaminants. In such a case, the problem is not necessarily the quantity of water but whether the water satisfies the requirements of the activity for which it is needed.

We can therefore distinguish accessible water from usable water:

$$
W_{\text{usable}}(t)\leq W_{\text{accessible}}(t)
$$

A polluted river may contain abundant water while providing very little usable water to the agent. Water treatment changes this relationship. Filtration, disinfection, desalination, and other processes can transform water that was previously unusable into water that satisfies the agent's requirements.

A sufficient supply of usable water may still fail to solve the problem if the agent cannot depend upon it. A source may provide sufficient water under normal conditions but fail during a drought. Rainfall may provide abundant water during one season and almost none during another. A municipal system may normally provide reliable service but remain vulnerable to infrastructure failure.

The relevant issue is therefore not merely whether sufficient water is available at a particular moment, but whether the agent can expect sufficient access across the period relevant to its activities.

If $E$ represents the event that sufficient water will be available when required, then one possible measure of reliability is:

$$
R(t)=P(E\mid t)
$$

A source with high expected availability but low reliability can therefore present a serious water problem. Storage is one way of addressing this problem. A reservoir, cistern, or household water tank allows water obtained during periods of abundance to support activities during periods when the source is unavailable.

Water can also be available in sufficient quantity and be usable, yet still be unavailable where it is needed. A farming region may be located far from a reliable water source. The problem in such a case is not necessarily a shortage of water in the environment. Rather, water and the activity requiring it occupy different locations.

The distribution of water can therefore be represented as a function of both location and time:

$$
W(x,t)
$$

An agent's requirements are likewise associated with particular locations. A crop requires water at the location of the field. A household requires water at the location of the household. A city requires water within its distribution system. The water problem consequently depends upon the relationship between the location of supply and the location of demand.

Canals, aqueducts, pipelines, and transportation systems address this form of the problem by moving water through space.

The same problem can occur through time. Water may be available at a time when it is not needed and unavailable when it is needed. Rainfall may occur during a season when crops require little water and disappear during the growing season when demand is greatest. A river may have abundant flow during the spring but insufficient flow during the summer.

The problem is therefore not necessarily the total quantity of water available, but the relationship between water supply and water requirements through time:

$$
W_{\text{available}}(t)
\quad\text{and}\quad
W_{\text{required}}(t)
$$

Storage once again provides a means of overcoming the constraint. A reservoir effectively moves water through time, allowing water acquired during periods of abundance to satisfy requirements during periods of scarcity.

Finally, water may be physically accessible and usable while still being too costly to obtain. Water extraction may require pumps and electricity. Water may need to be transported over long distances. Contaminated water may require expensive treatment. A household may have access to a municipal system but lack the resources necessary to purchase sufficient water.

The relevant question is therefore not simply whether water can be obtained, but whether it can be obtained using resources the agent is capable of committing to the problem.

If $C_W(t)$ represents the cost of obtaining the required water and $B(t)$ represents the resources available to the agent for that purpose, then a necessary condition may be expressed as:

$$
C_W(t)\leq B(t)
$$

A source can therefore be physically accessible while remaining practically inaccessible because its cost exceeds the resources available to the agent.

These forms of the water problem should not yet be interpreted as independent dimensions. Indeed, they frequently interact. A drought can reduce both availability and reliability. Pollution can reduce usability while increasing the cost of obtaining usable water. Distance can increase transportation costs. Lack of storage can turn a temporary shortage into a recurring problem. Increased energy costs can make previously accessible groundwater effectively inaccessible.

The forms of the water problem are therefore best understood initially as different ways in which the basic constraint can fail. The underlying problem remains:

$$
\boxed{
\text{maintaining sufficient access to usable water over time}
}
$$

What varies is the reason that this condition cannot be maintained. The obstacle may be insufficient quantity, inability to reach the available supply, inadequate quality, uncertainty about future supply, separation between supply and demand in space or time, or the resources required to obtain the water.

This distinction is important because the different forms need not correspond to different kinds of information. Some may ultimately prove to be consequences of other facts about the agent and its environment. For example, if the location of a water source, the agent's technological capabilities, and its available resources are already known, perhaps its accessibility can be determined from those facts. Likewise, reliability may be derivable from a sufficiently complete description of the source and its environment.

Alternatively, some of these properties may contain information that cannot be recovered from the others.

Determining which possibility is correct is not primarily a question about water. It is a question about the structure of **position** itself.

The purpose of this article is therefore not to declare that availability, accessibility, quality, reliability, spatial mismatch, temporal mismatch, and cost are separate dimensions. It is to identify the structure of the water problem clearly enough that the later theoretical analysis can determine what information an agent actually needs.

The next question is therefore how deeply water enters human activity. Water is not required merely to keep an individual alive. It enters agriculture, settlement, sanitation, industry, energy, transportation, and the productive capacity of entire societies. The water problem consequently extends far beyond the simple question of obtaining something to drink.


## Water as a Constraint on Action
The water problem becomes more significant when we recognize that water is not merely a requirement for biological survival. Water can determine what an agent is able to do.

An individual requires water to remain alive, but human activity extends far beyond survival. People cultivate crops, raise animals, build settlements, manufacture goods, generate energy, transport materials, and perform countless other activities that depend directly or indirectly upon access to water. When water is unavailable, inaccessible, unusable, or too costly to obtain, these activities may become more difficult or impossible.

Water therefore acts as a constraint on action.

Suppose an agent has a set of actions that it could otherwise perform. Some of those actions will require water. If the agent cannot satisfy the relevant water requirement, those actions may no longer be feasible. The agent may possess the necessary knowledge, labor, equipment, and other resources, but the absence of sufficient usable water can still prevent the action from occurring.

We can express this relationship generally by considering the set of actions available to an agent:

$$
A(P)
$$

where $P$ represents the agent's current position. If an action $a$ requires sufficient access to water, then a water constraint can determine whether $a$ belongs to the feasible set of actions.

In this sense, water does not need to be the objective of an action in order to constrain it. A farmer does not necessarily want water for its own sake. The farmer wants to produce a crop, but producing the crop may require water. A factory may not seek water as its final objective, but its production process may depend upon a sufficient supply. A settlement may seek to grow, but its population and productive activities may be constrained by the amount of water its infrastructure can reliably supply.

The water problem therefore propagates through human activity. A constraint on water can become a constraint on food production, which can become a constraint on population, which can become a constraint on settlement and economic activity.

This relationship can also operate in the opposite direction. An activity can increase the amount of water required by an agent. Cultivating more land, supporting a larger population, or establishing a water-intensive industry can increase $W_{\text{required}}(t)$. The agent's actions therefore change the water problem that the agent subsequently faces.

The constraint is consequently dynamic. An agent acts within a particular relationship to water, and those actions can alter that relationship:

$$
P\rightarrow P'
$$

The resulting position may contain greater or lesser access to water, greater or lesser requirements, or greater or lesser ability to obtain water in the future.

Consider agriculture. A farmer with reliable rainfall may be able to cultivate land without constructing an irrigation system. A farmer in a dry region may require wells, canals, reservoirs, or pumps before the same agricultural activity becomes possible. The difference is not necessarily the farmer's knowledge of agriculture or the fertility of the land. It may be the farmer's relationship to water.

The same principle applies to settlement. A location with a nearby and reliable water source can support activities that would be difficult or impossible at a location where water must be transported over great distances. As settlements grow, however, their water requirements also grow. A water source that was sufficient for a small settlement may become insufficient for a city.

Water can therefore influence not only which actions are possible but also where those actions can be performed and at what scale.

This helps explain why human beings have repeatedly altered their physical environment in order to change their relationship to water. Wells allow agents to access groundwater. Irrigation systems deliver water to agricultural land. Reservoirs store water for later use. Pipelines move water toward population centers. Treatment systems convert contaminated water into usable water.

These technologies do more than provide water. They change the set of actions that can be performed from a given location.

A previously uninhabitable location may become suitable for settlement. Previously unproductive land may become agricultural. A city may be able to support a larger population. An industrial process that was previously infeasible may become possible.

The water problem therefore has an important relationship to opportunity. A limitation on water can remove actions from the set of feasible choices, while an improvement in access to water can add actions to that set.

This does not mean that water determines human behavior. Agents can respond to water constraints in many ways. They can reduce their requirements, substitute other activities, acquire new sources, build infrastructure, store water, move water, or move themselves. The constraint affects the choices available to the agent, but it does not by itself determine which choice the agent makes.

The distinction is important. A constraint on action is not the same thing as a command to act.

An agent facing a water shortage might build a well. It might conserve water. It might purchase water from another source. It might abandon a water-intensive activity. It might relocate. Or it might accept the consequences of the shortage. The water problem changes the decision problem by changing what is feasible and what each possible response requires.

Water therefore enters human decision-making not only as a resource to be acquired, but as a condition underlying other actions.

This gives the water problem a broader structure. The agent must maintain sufficient access to usable water, but the purpose of doing so is not merely to accumulate water. Water is valuable because it enables the agent to sustain and perform activities that depend upon it.

The problem can therefore be understood as a relationship between water and action:

$$
\text{Water Constraint}
\rightarrow
\text{Feasible Actions}
\rightarrow
\text{Possible Outcomes}
$$

Changing the water constraint can change the actions available to the agent, which can in turn change the outcomes the agent can reach.

This is why solving the water problem is not simply a matter of finding more water. It is a matter of changing the conditions under which an agent can act.

The next question is therefore how agents accomplish this. If water constrains action, what can an agent do to change that constraint? Human history contains an enormous variety of answers: wells, reservoirs, canals, aqueducts, pipelines, treatment systems, irrigation, storage, recycling, and many others.

These solutions can be understood not merely as technologies for obtaining water, but as different ways of transforming the relationship between an agent and the water constraint.
