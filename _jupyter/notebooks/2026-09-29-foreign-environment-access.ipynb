## Introduction
An agent does not necessarily need to change its current environment in order to benefit from another environment. It may instead seek to acquire access to environments that it does not currently occupy.

This possibility creates a particular problem within the broader environment problem. An agent may exist within an environment $E_0$ while other environments $E_1,E_2,\ldots,E_n$ remain outside its current reach. The question is therefore not simply how an agent acts within its environment, but how it can establish a relationship with environments beyond the one it currently occupies.

Let the set of environments available to an agent from position $P$ be represented by

$$
\mathcal{A}_E(P)\subseteq\mathcal{E}.
$$

For a foreign environment $E_i$, the problem of access can therefore be represented initially as a transition:

$$
E_i\notin\mathcal{A}_E(P)
\rightarrow
E_i\in\mathcal{A}_E(P).
$$

This representation is intentionally simple. It does not yet specify what it means for an environment to be accessible. An agent may be able to enter an environment without being able to remain there, reside there, work there, establish a business there, or return to it repeatedly. The relationship between an agent and an environment may therefore be more complex than a binary distinction between access and non-access.

This raises a fundamental question:

> **What does it mean for an agent to have access to a foreign environment?**

Answering this question requires examining the conditions under which an agent can enter, remain within, operate within, and maintain a continuing relationship with an environment outside its current one.

These conditions can differ substantially between environments. An agent's access may depend on legal authorization, financial resources, physical mobility, knowledge, relationships, time, or other properties of the agent and its circumstances. The relative importance of these conditions cannot, however, be assumed in advance. They must be identified by examining the actual mechanisms through which access is acquired and maintained.

Legal status provides one particularly visible mechanism. A visa, residence permit, or other authorization may change the activities an agent can perform within an environment and the length of time it can remain there. But legal authorization is only one component of environmental access. An agent may possess legal permission while lacking the resources or capabilities required to exercise that permission effectively.

The distinction can therefore be made between the existence of an environment, the legal ability to enter it, and the practical ability to make use of it:

$$
\text{Environment}
\neq
\text{Legal Access}
\neq
\text{Effective Access}.
$$

The problem becomes even more interesting when an agent can maintain access to multiple environments simultaneously. An agent may occupy $E_0$ while retaining access to $E_1$ and $E_2$:

$$
E_0\quad\text{and}\quad
\mathcal{A}_E(P)=\{E_0,E_1,E_2\}.
$$

In this case, the agent's environmental circumstances are not determined solely by the environment it currently occupies. Its future decision space may also depend on the environments it can enter if circumstances change.

Foreign environmental access therefore provides a concrete setting in which to investigate a broader question from the environment problem: **how does an agent's relationship with environments outside its current environment affect the positions it can reach?**

The objective of this investigation is not to define a final representation of environmental access in advance. It is to examine the mechanisms through which foreign environmental access is acquired, maintained, exercised, and expanded, and to determine what these mechanisms reveal about the relationship between an agent, its environment, and its future possibilities.

The investigation begins with the simplest case: an agent seeking access to an environment it does not currently occupy.
