## The Food Problem
Every agent that persists through time must solve a collection of fundamental problems. These problems arise from the conditions under which the agent exists and from the requirements imposed by its continued existence. An agent cannot simply choose whether to solve them. It must either solve them, or eventually cease to occupy the position from which it can act.

Food is one such problem.

At first glance, the food problem appears simple: an organism must eat. But this description conceals the underlying problem. An organism is continuously consuming matter and energy in order to maintain its physical organization. What it has available at one moment is therefore gradually depleted. To continue existing, it must somehow replenish what it consumes.

The problem is consequently not merely obtaining food once. It is maintaining sufficient access to nourishment through time.

This distinction is important because the availability of food at the present moment does not determine whether an agent has solved the problem. An individual may possess enough food for today while having no reliable means of obtaining food tomorrow. Another may possess little food at present while having secure access to a productive food system. These agents occupy different positions with respect to the same underlying problem.

The food problem is therefore fundamentally a problem of **continued access**. An agent must maintain the capacity to acquire sufficient usable nourishment as its existing resources are consumed and as the conditions surrounding its future access change.

This immediately makes the problem strategic. Food must be acquired, but it may also be produced, stored, preserved, exchanged, transported, or obtained through cooperation with other agents. Different organisms and societies employ radically different strategies, yet these strategies address the same underlying requirement: maintaining the resources necessary for continued biological activity.

The purpose of this article is to investigate that requirement formally. We will begin with the physical necessity of metabolic maintenance, examine why consumption creates a problem of replenishment, and then consider why that problem necessarily extends into the future. From there, we can identify the conditions under which an agent can be said to have a secure or insecure food position.

Only after establishing the problem itself should we attempt to define the **Food Dimension of Position**. The dimension should not be imposed upon the theory in advance. It should emerge from the problem: from the information about an agent's position that determines its ability to solve the food problem and continue reaching viable future positions.
